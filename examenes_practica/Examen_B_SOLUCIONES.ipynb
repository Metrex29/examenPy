## Sección 3 · Empleados funcional

Trabaja con:
```python
empleados = [
    {"nombre": "Ana",   "salario": 1800, "activo": True},
    {"nombre": "Luis",  "salario": 1200, "activo": False},
    {"nombre": "Marta", "salario": 2500, "activo": True},
    {"nombre": "Hugo",  "salario": 900,  "activo": True},
]
```
1. Con `sorted()` y una lambda, los nombres ordenados de **mayor a menor** salario.
2. Con `filter()`, los nombres de los empleados activos.
3. Con `reduce()`, el salario más alto de todos.

````python
from functools import reduce
empleados = [
    {"nombre": "Ana",   "salario": 1800, "activo": True},
    {"nombre": "Luis",  "salario": 1200, "activo": False},
    {"nombre": "Marta", "salario": 2500, "activo": True},
    {"nombre": "Hugo",  "salario": 900,  "activo": True},
]
print([e["nombre"] for e in sorted(empleados, key=lambda e: e["salario"], reverse=True)])
print([e["nombre"] for e in filter(lambda e: e["activo"], empleados)])
print(reduce(lambda a, b: a if a["salario"] >= b["salario"] else b, empleados)["salario"])
````

## Sección 4 · Contar vocales

Escribe una función recursiva `contar_vocales(texto)` que devuelva cuántas vocales hay, **sin bucles**. Pista: mira el primer carácter y suma 1 si es vocal, luego sigue con el resto.

```python
print(contar_vocales("murcielago"))
print(contar_vocales("xyz"))
```

````python
def contar_vocales(texto):
    if texto == "":
        return 0
    primero = 1 if texto[0] in "aeiou" else 0
    return primero + contar_vocales(texto[1:])
print(contar_vocales("murcielago"))
print(contar_vocales("xyz"))
````

## Sección 5 · Clase CuentaPuntos

Crea una clase `CuentaPuntos` con:
- atributo de clase `total_jugadores` que cuente cuántas se crean;
- atributos `jugador` y `_puntos` (0 por defecto);
- método `sumar(p)` que sume puntos solo si `p` es positivo;
- método `__str__` con el formato `Ana: 30 puntos`.

```python
c1 = CuentaPuntos("Ana")
c2 = CuentaPuntos("Leo")
c1.sumar(30)
c1.sumar(-5)
print(c1)
print(c2)
print("Jugadores:", CuentaPuntos.total_jugadores)
```

````python
class CuentaPuntos:
    total_jugadores = 0
    def __init__(self, jugador, puntos=0):
        self.jugador = jugador
        self._puntos = puntos
        CuentaPuntos.total_jugadores += 1
    def sumar(self, p):
        if p > 0:
            self._puntos += p
    def __str__(self):
        return f"{self.jugador}: {self._puntos} puntos"
c1 = CuentaPuntos("Ana")
c2 = CuentaPuntos("Leo")
c1.sumar(30)
c1.sumar(-5)
print(c1)
print(c2)
print("Jugadores:", CuentaPuntos.total_jugadores)
````

## Sección 6 · Clase Cesta

Crea una clase `Cesta` con un `nombre` y una lista interna de productos. Haz que funcionen:
- `len(cesta)` → número de productos;
- `"X" in cesta` → si un producto está;
- `cesta += "X"` → añade y devuelve la propia cesta;
- `str(cesta)` → `Compra (2 productos)`.

```python
c = Cesta("Compra")
c += "pan"
c += "leche"
print(c)
print(len(c))
print("pan" in c)
print("huevos" in c)
```

````python
class Cesta:
    def __init__(self, nombre):
        self.nombre = nombre
        self._productos = []
    def __len__(self):
        return len(self._productos)
    def __contains__(self, x):
        return x in self._productos
    def __iadd__(self, x):
        self._productos.append(x)
        return self
    def __str__(self):
        return f"{self.nombre} ({len(self)} productos)"
c = Cesta("Compra")
c += "pan"
c += "leche"
print(c)
print(len(c))
print("pan" in c)
print("huevos" in c)
````

## Sección 7 · Animales (polimorfismo)

1. Crea `Animal` con `nombre` y un método `sonido()` que devuelva `"..."`.
2. `Perro` devuelve `"Guau"`, `Gato` devuelve `"Miau"`.
3. `Cachorro` hereda de `Perro` y su `sonido()` es el del padre más `" (flojito)"` (usa `super()`).
4. Recorre una lista de animales mostrando `nombre: sonido`.

```python
animales = [Perro("Rex"), Gato("Tom"), Cachorro("Boti")]
```

````python
class Animal:
    def __init__(self, nombre):
        self.nombre = nombre
    def sonido(self):
        return "..."
class Perro(Animal):
    def sonido(self):
        return "Guau"
class Gato(Animal):
    def sonido(self):
        return "Miau"
class Cachorro(Perro):
    def sonido(self):
        return super().sonido() + " (flojito)"
animales = [Perro("Rex"), Gato("Tom"), Cachorro("Boti")]
for a in animales:
    print(f"{a.nombre}: {a.sonido()}")
````

## Sección 8 · Transporte (refactor)

Este código usa `if/elif` por tipo:
```python
viajes = [["taxi", 10], ["bus", 10], ["metro", 10]]
for v in viajes:
    if v[0] == "taxi":
        precio = v[1] * 1.5 + 2
    elif v[0] == "bus":
        precio = v[1] * 0.5
    elif v[0] == "metro":
        precio = v[1] * 0.4
    print(v[0], round(precio, 2))
```
Reescríbelo con una clase `Viaje` y subclases `Taxi`, `Bus` y `Metro`, cada una con `precio()`. Los km se pasan al crear el objeto. Sin `if` de tipo.

```python
viajes = [Taxi(10), Bus(10), Metro(10)]
for v in viajes:
    print(type(v).__name__, round(v.precio(), 2))
```

````python
class Viaje:
    def __init__(self, km):
        self.km = km
    def precio(self):
        return self.km
class Taxi(Viaje):
    def precio(self):
        return self.km * 1.5 + 2
class Bus(Viaje):
    def precio(self):
        return self.km * 0.5
class Metro(Viaje):
    def precio(self):
        return self.km * 0.4
viajes = [Taxi(10), Bus(10), Metro(10)]
for v in viajes:
    print(type(v).__name__, round(v.precio(), 2))
````

## Sección 9 · Gimnasio

Clase `Socio` con `nombre`, `edad` y `cuota`. Con:
```python
socios = [
    Socio("Ana", 30, 40.0),
    Socio("Luis", 17, 25.0),
    Socio("Marta", 45, 50.0),
    Socio("Hugo", 16, 25.0),
]
```
1. Con un **set** comprehension, las edades distintas (ordénalas al mostrarlas).
2. Con `filter()`, los nombres de los socios mayores de edad (18 o más).
3. Función recursiva `ingresos(socios)` que sume las cuotas **sin bucles**.

````python
class Socio:
    def __init__(self, nombre, edad, cuota):
        self.nombre = nombre
        self.edad = edad
        self.cuota = cuota
socios = [Socio("Ana", 30, 40.0), Socio("Luis", 17, 25.0), Socio("Marta", 45, 50.0), Socio("Hugo", 16, 25.0)]
print(sorted({s.edad for s in socios}))
print([s.nombre for s in filter(lambda s: s.edad >= 18, socios)])
def ingresos(socios):
    if not socios:
        return 0
    return socios[0].cuota + ingresos(socios[1:])
print(ingresos(socios))
````
