## Sección 3 · Ventas con reduce avanzado

```python
ventas = [
    {"vendedor": "Ana",  "producto": "A", "importe": 100, "devuelto": False},
    {"vendedor": "Luis", "producto": "B", "importe": 200, "devuelto": True},
    {"vendedor": "Ana",  "producto": "C", "importe": 150, "devuelto": False},
    {"vendedor": "Luis", "producto": "A", "importe": 300, "devuelto": False},
]
```
1. Con `filter()` y `reduce()`, el importe total de las ventas **no devueltas**.
2. Con `map()`, una lista de textos `"Ana vendió A por 100€"`.
3. Con `reduce()`, construye un diccionario `{vendedor: importe acumulado}` contando **solo las no devueltas**. Pista: el acumulador es un diccionario y empieza en `{}`.

````python
from functools import reduce
ventas = [
    {"vendedor": "Ana",  "producto": "A", "importe": 100, "devuelto": False},
    {"vendedor": "Luis", "producto": "B", "importe": 200, "devuelto": True},
    {"vendedor": "Ana",  "producto": "C", "importe": 150, "devuelto": False},
    {"vendedor": "Luis", "producto": "A", "importe": 300, "devuelto": False},
]
print(reduce(lambda a, v: a + v["importe"], filter(lambda v: not v["devuelto"], ventas), 0))
print(list(map(lambda v: f'{v["vendedor"]} vendió {v["producto"]} por {v["importe"]}€', ventas)))
def acumular(d, v):
    if not v["devuelto"]:
        d[v["vendedor"]] = d.get(v["vendedor"], 0) + v["importe"]
    return d
print(reduce(acumular, ventas, {}))
````

## Sección 4 · Aplanar listas anidadas (recursivo)

Escribe una función recursiva `aplanar(lista)` que convierta una lista con listas dentro en una lista plana. Puedes usar un `for` para recorrer, pero las sublistas se tratan con la llamada recursiva. Pista: `isinstance(x, list)`.
```python
print(aplanar([1, [2, 3], [[4], 5], [], 6]))
print(suma_profunda([1, [2, [3, [4]]]]))
```
Y una `suma_profunda(lista)` que sume todos los números por muy anidados que estén.

````python
def aplanar(lista):
    resultado = []
    for x in lista:
        if isinstance(x, list):
            resultado += aplanar(x)
        else:
            resultado.append(x)
    return resultado
def suma_profunda(lista):
    total = 0
    for x in lista:
        total += suma_profunda(x) if isinstance(x, list) else x
    return total
print(aplanar([1, [2, 3], [[4], 5], [], 6]))
print(suma_profunda([1, [2, [3, [4]]]]))
````

## Sección 5 · Clase Inventario (composición + classmethod)

Crea una clase `Producto` con `nombre`, `precio` y `stock`. Luego una clase `Inventario` que **contenga** una lista de productos con:
- `añadir(producto)`;
- `valor_total()` → suma de `precio × stock` de todos;
- `sin_stock()` → lista de nombres con stock 0;
- método de clase `desde_pares(pares)` que reciba una lista de tuplas `(nombre, precio, stock)` y devuelva un `Inventario` ya montado (usa `@classmethod`).

```python
inv = Inventario.desde_pares([("Pan", 1.0, 10), ("Leche", 0.8, 0), ("Huevos", 2.0, 5)])
print(inv.valor_total())
print(inv.sin_stock())
```

````python
class Producto:
    def __init__(self, nombre, precio, stock):
        self.nombre = nombre
        self.precio = precio
        self.stock = stock
class Inventario:
    def __init__(self):
        self.productos = []
    def añadir(self, producto):
        self.productos.append(producto)
    def valor_total(self):
        return sum(p.precio * p.stock for p in self.productos)
    def sin_stock(self):
        return [p.nombre for p in self.productos if p.stock == 0]
    @classmethod
    def desde_pares(cls, pares):
        inv = cls()
        for nombre, precio, stock in pares:
            inv.añadir(Producto(nombre, precio, stock))
        return inv
inv = Inventario.desde_pares([("Pan", 1.0, 10), ("Leche", 0.8, 0), ("Huevos", 2.0, 5)])
print(inv.valor_total())
print(inv.sin_stock())
````

## Sección 6 · Clase Vector (operadores)

Crea una clase `Vector` con `x` e `y`. Haz que funcionen:
- `+` → nuevo `Vector` sumando componente a componente;
- `*` por un número → nuevo `Vector` con cada componente multiplicada;
- `==` → compara ambas componentes;
- `len(v)` → el módulo redondeado a entero (`int((x**2 + y**2) ** 0.5)`);
- `str(v)` → `(3, 4)`.

```python
a = Vector(3, 4)
b = Vector(1, 2)
print(a + b)
print(a * 2)
print(a == Vector(3, 4))
print(len(a))
```

````python
class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y
    def __add__(self, o):
        return Vector(self.x + o.x, self.y + o.y)
    def __mul__(self, n):
        return Vector(self.x * n, self.y * n)
    def __eq__(self, o):
        return self.x == o.x and self.y == o.y
    def __len__(self):
        return int((self.x ** 2 + self.y ** 2) ** 0.5)
    def __str__(self):
        return f"({self.x}, {self.y})"
a = Vector(3, 4)
b = Vector(1, 2)
print(a + b)
print(a * 2)
print(a == Vector(3, 4))
print(len(a))
````

Refactoriza con clases y deja preparado para ampliar. Superclase `Plan` con `cliente` y `base`, método `coste()` que devuelve `base`, y `__str__` → `Básico: Ana paga 10.00 €` (el tipo con `type(self).__name__`). Subclases:
- `Premium`: cuesta el doble de la base;
- `Pro`: base + 5 € fijos.
Crea una clase `Cartera` que contenga los planes, con `total()` (suma de costes) y `__str__` que liste todos. Luego **sin tocar nada**, añade un plan `Gratis` que cueste 0.

```python
cartera = Cartera()
cartera.añadir(Plan("Ana", 10))
cartera.añadir(Premium("Luis", 10))
cartera.añadir(Pro("Marta", 10))
cartera.añadir(Gratis("Hugo", 10))

print(f"Total: {cartera.total():.2f} €")
```

````python
class Plan:
    def __init__(self, cliente, base):
        self.cliente = cliente
        self.base = base
    def coste(self):
        return self.base
    def __str__(self):
        return f"{type(self).__name__}: {self.cliente} paga {self.coste():.2f} €"
class Premium(Plan):
    def coste(self):
        return self.base * 2
class Pro(Plan):
    def coste(self):
        return self.base + 5
class Gratis(Plan):
    def coste(self):
        return 0
class Cartera:
    def __init__(self):
        self.planes = []
    def añadir(self, plan):
        self.planes.append(plan)
    def total(self):
        return sum(p.coste() for p in self.planes)
    def __str__(self):
        return "\n".join(str(p) for p in self.planes)
cartera = Cartera()
cartera.añadir(Plan("Ana", 10))
cartera.añadir(Premium("Luis", 10))
cartera.añadir(Pro("Marta", 10))
cartera.añadir(Gratis("Hugo", 10))
print(cartera)
print(f"Total: {cartera.total():.2f} €")
````

## Sección 9 · Pedidos de una cafetería (integrador)

```python
class Bebida:
    def __init__(self, nombre, precio):
        self.nombre = nombre
        self.precio = precio
    def precio_final(self):
        return self.precio

class ConLeche(Bebida):      # +0.50 €
    def precio_final(self):
        return super().precio_final() + 0.50

class Grande(Bebida):        # +30 % sobre el precio
    def precio_final(self):
        return super().precio_final() * 1.30
```
Con esta lista:
```python
pedidos = [
    {"bebida": ConLeche("Café", 1.20), "uds": 2},
    {"bebida": Bebida("Agua", 1.00),   "uds": 1},
    {"bebida": Grande("Zumo", 2.00),   "uds": 3},
]
```
1. Con una comprehension, lista con el importe de cada pedido (`precio_final() × uds`) redondeado a 2.
2. Con `map()` + `reduce()`, la facturación total.
3. Con `filter()` e `isinstance()`, los nombres de las bebidas `ConLeche`.
4. Recursiva `total_uds(pedidos)` que sume las unidades sin bucles.

````python
from functools import reduce
class Bebida:
    def __init__(self, nombre, precio):
        self.nombre = nombre
        self.precio = precio
    def precio_final(self):
        return self.precio
class ConLeche(Bebida):
    def precio_final(self):
        return super().precio_final() + 0.50
class Grande(Bebida):
    def precio_final(self):
        return super().precio_final() * 1.30
pedidos = [
    {"bebida": ConLeche("Café", 1.20), "uds": 2},
    {"bebida": Bebida("Agua", 1.00),   "uds": 1},
    {"bebida": Grande("Zumo", 2.00),   "uds": 3},
]
imp = lambda p: p["bebida"].precio_final() * p["uds"]
print([round(imp(p), 2) for p in pedidos])
print(round(reduce(lambda a, b: a + b, map(imp, pedidos), 0), 2))
print([p["bebida"].nombre for p in filter(lambda p: isinstance(p["bebida"], ConLeche), pedidos)])
def total_uds(pedidos):
    if not pedidos:
        return 0
    return pedidos[0]["uds"] + total_uds(pedidos[1:])
print(total_uds(pedidos))
````
