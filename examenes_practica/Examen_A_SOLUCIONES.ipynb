## Sección 3 · Tienda con map, filter y reduce

Trabaja con esta lista:
```python
productos = [
    {"nombre": "Teclado", "precio": 20.0, "oferta": True},
    {"nombre": "Ratón",   "precio": 12.0, "oferta": False},
    {"nombre": "Monitor", "precio": 150.0, "oferta": True},
    {"nombre": "Cable",   "precio": 5.0,  "oferta": False},
]
```
1. Con `map()`, una lista con los nombres en mayúsculas.
2. Con `filter()`, los nombres de los productos en oferta.
3. Con `reduce()`, el precio total de todos los productos.

````python
from functools import reduce
productos = [
    {"nombre": "Teclado", "precio": 20.0, "oferta": True},
    {"nombre": "Ratón",   "precio": 12.0, "oferta": False},
    {"nombre": "Monitor", "precio": 150.0, "oferta": True},
    {"nombre": "Cable",   "precio": 5.0,  "oferta": False},
]
print(list(map(lambda p: p["nombre"].upper(), productos)))
print([p["nombre"] for p in filter(lambda p: p["oferta"], productos)])
print(reduce(lambda acum, p: acum + p["precio"], productos, 0))
````

## Sección 4 · Potencia recursiva

Escribe una función recursiva `potencia(base, exp)` que calcule `base` elevado a `exp` **sin usar `**` ni bucles**. Caso base: exponente 0 devuelve 1.

```python
print(potencia(2, 5))
print(potencia(10, 0))
```

````python
def potencia(base, exp):
    if exp == 0:
        return 1
    return base * potencia(base, exp - 1)
print(potencia(2, 5))
print(potencia(10, 0))
````

## Sección 5 · Clase Pelicula

Crea una clase `Pelicula` con:
- atributos `titulo`, `duracion` (minutos) y `vista` (por defecto `False`);
- método `marcar_vista()` que ponga `vista` a `True`;
- método `es_larga()` que devuelva `True` si dura más de 120 minutos;
- `__str__` con el formato `Origen (148 min) [pendiente]` o `[vista]` según corresponda.

```python
p = Pelicula("Origen", 148)
print(p)
print(p.es_larga())
p.marcar_vista()
print(p)
```

````python
class Pelicula:
    def __init__(self, titulo, duracion, vista=False):
        self.titulo = titulo
        self.duracion = duracion
        self.vista = vista
    def marcar_vista(self):
        self.vista = True
    def es_larga(self):
        return self.duracion > 120
    def __str__(self):
        estado = "vista" if self.vista else "pendiente"
        return f"{self.titulo} ({self.duracion} min) [{estado}]"
p = Pelicula("Origen", 148)
print(p)
print(p.es_larga())
p.marcar_vista()
print(p)
````

## Sección 6 · Clase Temperatura

Crea una clase `Temperatura` que guarde `grados`. Haz que funcionen:
- `+` entre dos temperaturas → un **nuevo** `Temperatura`;
- `<` que compare los grados;
- `str()` con el formato `21.5 °C` (1 decimal);
- `float()` que devuelva los grados.

```python
a = Temperatura(21.5)
b = Temperatura(3.2)
print(a + b)
print(a < b)
print(float(a))
```

````python
class Temperatura:
    def __init__(self, grados):
        self.grados = grados
    def __add__(self, otra):
        return Temperatura(self.grados + otra.grados)
    def __lt__(self, otra):
        return self.grados < otra.grados
    def __str__(self):
        return f"{self.grados:.1f} °C"
    def __float__(self):
        return float(self.grados)
a = Temperatura(21.5)
b = Temperatura(3.2)
print(a + b)
print(a < b)
print(float(a))
````

## Sección 7 · Empleados

1. Crea `Empleado` con `nombre` y `sueldo_base`, y un método `sueldo()` que devuelva el sueldo base.
2. `Comercial` (hereda de `Empleado`): recibe además `ventas` y suma 0.10 € por cada venta al sueldo base (usa `super()`).
3. `Jefe` (hereda de `Empleado`): cobra el sueldo base más un 20 %.
4. Recorre una lista con los tres tipos mostrando `nombre: sueldo` (2 decimales) y el total.

```python
plantilla = [
    Empleado("Ana", 1000),
    Comercial("Luis", 1000, 50),
    Jefe("Marta", 1000),
]
```

````python
class Empleado:
    def __init__(self, nombre, sueldo_base):
        self.nombre = nombre
        self.sueldo_base = sueldo_base
    def sueldo(self):
        return self.sueldo_base
class Comercial(Empleado):
    def __init__(self, nombre, sueldo_base, ventas):
        super().__init__(nombre, sueldo_base)
        self.ventas = ventas
    def sueldo(self):
        return super().sueldo() + 0.10 * self.ventas
class Jefe(Empleado):
    def sueldo(self):
        return super().sueldo() * 1.20
plantilla = [Empleado("Ana", 1000), Comercial("Luis", 1000, 50), Jefe("Marta", 1000)]
total = 0
for e in plantilla:
    print(f"{e.nombre}: {e.sueldo():.2f}")
    total += e.sueldo()
print(f"Total: {total:.2f}")
````

## Sección 8 · Figuras geométricas

Este código usa `if/elif` por tipo:
```python
figuras = [["cuadrado", 4], ["circulo", 3], ["cuadrado", 2]]
for f in figuras:
    if f[0] == "cuadrado":
        area = f[1] * f[1]
    elif f[0] == "circulo":
        area = 3.14 * f[1] * f[1]
    print(f[0], round(area, 2))
```
Reescríbelo con una clase `Figura` y subclases `Cuadrado` y `Circulo`, cada una con un método `area()`, y recórrelo **sin ningún `if` de tipo**. El `lado`/`radio` se pasa al crear el objeto.

```python
figuras = [Cuadrado(4), Circulo(3), Cuadrado(2)]
for f in figuras:
    print(type(f).__name__, round(f.area(), 2))
```

````python
class Figura:
    def __init__(self, medida):
        self.medida = medida
    def area(self):
        return 0
class Cuadrado(Figura):
    def area(self):
        return self.medida * self.medida
class Circulo(Figura):
    def area(self):
        return 3.14 * self.medida * self.medida
figuras = [Cuadrado(4), Circulo(3), Cuadrado(2)]
for f in figuras:
    print(type(f).__name__, round(f.area(), 2))
````

## Sección 9 · Biblioteca

Clase `Libro` con `titulo`, `paginas` y `precio`. Con esta lista:
```python
libros = [
    Libro("Python", 300, 25.0),
    Libro("Historia", 500, 30.0),
    Libro("Cuento", 80, 10.0),
]
```
1. Con una comprehension, lista de títulos de libros de más de 100 páginas.
2. Con `map()` y `reduce()` (o `sum`), el precio total.
3. Función recursiva `paginas_totales(libros)` que sume las páginas **sin bucles**.

````python
from functools import reduce
class Libro:
    def __init__(self, titulo, paginas, precio):
        self.titulo = titulo
        self.paginas = paginas
        self.precio = precio
libros = [Libro("Python", 300, 25.0), Libro("Historia", 500, 30.0), Libro("Cuento", 80, 10.0)]
print([l.titulo for l in libros if l.paginas > 100])
print(reduce(lambda a, b: a + b, map(lambda l: l.precio, libros), 0))
def paginas_totales(libros):
    if not libros:
        return 0
    return libros[0].paginas + paginas_totales(libros[1:])
print(paginas_totales(libros))
````
