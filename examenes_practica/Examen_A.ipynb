## Sección 8 · Figuras geométricas

Este código usa `if/elif` por tipo:
```python
figuras = [["cuadrado", 4], ["circulo", 3], ["cuadrado", 2]]
for f in figuras:
    if f[0] == "cuadrado":
        area = f[1] * f[1]
    elif f[0] == "circulo":
        area = 3.14 * f[1] * f[1]
    print(f[0], round(area, 2))
```
Reescríbelo con una clase `Figura` y subclases `Cuadrado` y `Circulo`, cada una con un método `area()`, y recórrelo **sin ningún `if` de tipo**. El `lado`/`radio` se pasa al crear el objeto.
