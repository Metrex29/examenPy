## Sección 1. Lectura de código: ¿qué se muestra por pantalla?
### Ejercicio 1.1 · El contador de tiendas

````python
class Tienda:
    total_tiendas = 0

    def __init__(self, nombre):
        self.nombre = nombre
        Tienda.total_tiendas += 1

t1 = Tienda("Centro")
t2 = Tienda("Norte")
t1.total_tiendas = 10

print(Tienda.total_tiendas, t1.total_tiendas, t2.total_tiendas)
````

**Salida:** `2 10 2`

- Se crean 2 tiendas, así que el **atributo de clase** `Tienda.total_tiendas` vale 2.
- `t1.total_tiendas = 10` **no modifica** el atributo de clase: crea un **atributo de instancia nuevo** dentro de `t1` que "tapa" al de la clase.
- `t2` no tiene atributo propio, así que Python lo busca en la clase y encuentra el 2.

**Pregunta extra:** `t1` tiene su propio atributo de instancia (10) y `t2` sigue leyendo el de la clase (2).

### Ejercicio 1.2 · Dos listas que parecen tres

````python
pedidos_a = [101, 102, 103]
pedidos_b = pedidos_a
pedidos_c = pedidos_a[:]

pedidos_b.append(104)
pedidos_c.append(999)

print(pedidos_a)
print(pedidos_b)
print(pedidos_c)
print(pedidos_a is pedidos_b, pedidos_a is pedidos_c, pedidos_a == pedidos_b)
````

**Salida:**
```
[101, 102, 103, 104]
[101, 102, 103, 104]
[101, 102, 103, 999]
True False True
```
- `pedidos_b = pedidos_a` no copia: las dos variables **apuntan al mismo objeto**. Si cambias una, cambias "las dos".
- `pedidos_a[:]` crea una **copia nueva** (otro objeto), por eso `999` solo está en `pedidos_c`.
- `is` dice si son **el mismo objeto**; `==` dice si tienen **el mismo contenido**.

### Ejercicio 1.3 · Filtrar o clasificar

````python
valoraciones = [4, 7, 5, 9, 3]

resultado_1 = [v for v in valoraciones if v >= 5]
resultado_2 = ["POSITIVA" if v >= 5 else "NEGATIVA" for v in valoraciones]

print(resultado_1)
print(resultado_2)
````

**Salida:**
```
[7, 5, 9]
['NEGATIVA', 'POSITIVA', 'POSITIVA', 'POSITIVA', 'NEGATIVA']
```
**Pregunta extra:**
- `if` **al final** (después del `for`) es un **filtro**: decide qué elementos entran. No admite `else`. La lista puede quedar más corta.
- `if ... else` **al principio** (antes del `for`) es una **expresión condicional**: decide qué valor se pone para cada elemento. Necesita `else` y la lista tiene la misma longitud que la original.

### Ejercicio 1.4 · Cuenta atrás

````python
def cuenta_atras(n):
    if n == 0:
        return "¡Despegue!"
    return str(n) + " " + cuenta_atras(n - 1)

print(cuenta_atras(3))
````

**Salida:** `3 2 1 ¡Despegue!`

Se va construyendo: `"3 " + ("2 " + ("1 " + "¡Despegue!"))`.

**Pregunta extra:** sin el `if` no hay **caso base**: la función se llama a sí misma sin parar (3, 2, 1, 0, -1, -2…) hasta que Python supera el límite de llamadas y lanza **`RecursionError: maximum recursion depth exceeded`**. Lo comprobamos:

````python
def cuenta_atras_sin_base(n):
    return str(n) + " " + cuenta_atras_sin_base(n - 1)

try:
    cuenta_atras_sin_base(3)
except RecursionError as e:
    print("RecursionError:", e)
````

### Ejercicio 1.5 · Recepción de un hotel

````python
class Huesped:
    def bienvenida(self):
        return "Bienvenido"

class HuespedVIP(Huesped):
    def bienvenida(self):
        return super().bienvenida() + " - Tiene champán en la habitación"

class HuespedGrupo(Huesped):
    def bienvenida(self):
        return "Bienvenidos, el guía les espera en el vestíbulo"

for huesped in [Huesped(), HuespedVIP(), HuespedGrupo()]:
    print(huesped.bienvenida())
````

**Salida:**
```
Bienvenido
Bienvenido - Tiene champán en la habitación
Bienvenidos, el guía les espera en el vestíbulo
```
Es **polimorfismo**: la misma llamada `bienvenida()` hace cosas distintas según la clase del objeto. `HuespedVIP` **amplía** el método del padre con `super()`; `HuespedGrupo` lo **sustituye** completamente.

### Ejercicio 1.6 · Atributos con guiones bajos

````python
class Tarjeta:
    def __init__(self):
        self.titular = "Ana"
        self._saldo = 500
        self.__pin = 1234

t = Tarjeta()
print(t.titular)        # Línea A
print(t._saldo)         # Línea B
print(t._Tarjeta__pin)  # Línea C
try:
    print(t.__pin)      # Línea D
except AttributeError as e:
    print("AttributeError:", e)
````

**Salida:** `Ana`, `500`, `1234` y en la línea D un error.

- **A** funciona: atributo público.
- **B** funciona: un guion bajo `_saldo` es solo una **convención** ("no lo toques desde fuera"), Python no lo impide.
- **C** funciona: con dos guiones bajos Python aplica **name mangling** y renombra el atributo a `_Tarjeta__pin`.
- **D** falla con **`AttributeError`**: el atributo `__pin` "no existe" con ese nombre fuera de la clase.

*(He puesto la línea D dentro de un `try/except` para que el cuaderno siga ejecutándose.)*

### Ejercicio 1.7 · Funciones que fabrican funciones

````python
def crear_etiqueta(prefijo):
    def etiquetar(codigo):
        return prefijo + "-" + str(codigo)
    return etiquetar

factura = crear_etiqueta("FAC")
albaran = crear_etiqueta("ALB")

print(factura(15), albaran(7), factura(albaran(1)))
````

**Salida:** `FAC-15 ALB-7 FAC-ALB-1`

`crear_etiqueta` devuelve una función interna que **recuerda** el `prefijo` con el que se creó (esto se llama *closure*). En `factura(albaran(1))` primero se evalúa `albaran(1)` → `"ALB-1"` y luego `factura("ALB-1")` → `"FAC-ALB-1"`.

### Ejercicio 1.8 · La cafetería y la herencia múltiple

````python
class Bebida:
    def ingredientes(self):
        return ["café"]

class ConLeche(Bebida):
    def ingredientes(self):
        return super().ingredientes() + ["leche"]

class ConHielo(Bebida):
    def ingredientes(self):
        return super().ingredientes() + ["hielo"]

class Frappe(ConLeche, ConHielo):
    def ingredientes(self):
        return super().ingredientes() + ["nata"]

class CafeFrio(ConHielo, ConLeche):
    pass

print(Frappe().ingredientes())
print(CafeFrio().ingredientes())
print([clase.__name__ for clase in Frappe.__mro__])
````

**Salida:**
```
['café', 'hielo', 'leche', 'nata']
['café', 'leche', 'hielo']
['Frappe', 'ConLeche', 'ConHielo', 'Bebida', 'object']
```
**Preguntas extra:**
1. `super()` **no significa "mi padre"**, significa "**la siguiente clase en el MRO del objeto**". En un `Frappe` el MRO es `Frappe → ConLeche → ConHielo → Bebida → object`, así que el `super()` de `ConLeche` llama a **`ConHielo`**.
2. El MRO pone cada clase **una sola vez** (y `Bebida` al final, después de todos sus hijos). Como la cadena de `super()` recorre el MRO, `Bebida.ingredientes()` solo se ejecuta una vez.
3. En `CafeFrio(ConHielo, ConLeche)` el MRO es `CafeFrio → ConHielo → ConLeche → Bebida`. Los ingredientes se añaden "a la vuelta" de las llamadas: primero se resuelve `Bebida` (café), luego `ConLeche` añade leche y por último `ConHielo` añade hielo. Al cambiar el orden de los padres, cambia el MRO y, con él, el orden.

````python
butacas_libres = {"Sala 1": 0, "Sala 2": 34, "Sala 3": 6, "Sala 4": 0, "Sala 5": 120}

# 1. Salas completas
completas = [sala for sala, libres in butacas_libres.items() if libres == 0]

# 2. Estado de cada sala (if/else encadenados en la expresión)
estados = {
    sala: "COMPLETA" if libres == 0 else "ÚLTIMAS PLAZAS" if libres < 10 else "DISPONIBLE"
    for sala, libres in butacas_libres.items()
}

# 3. Porcentaje de ocupación de las salas no completas
CAPACIDAD = 150
ocupacion = {
    sala: round((CAPACIDAD - libres) / CAPACIDAD * 100, 1)
    for sala, libres in butacas_libres.items() if libres > 0
}

print(completas)
print(estados)
print(ocupacion)
````

### Ejercicio 3.4 · `reduce()`

````python
# 1. Recaudación total: acumulador que empieza en 0
total = reduce(lambda acum, c: acum + c["importe"], carreras, 0)

# 2. Carrera más larga: sin valor inicial; nos quedamos con la mayor de cada pareja
mas_larga = reduce(lambda a, b: a if a["km"] >= b["km"] else b, carreras)

# 3. Km pagados con tarjeta: primero filter, luego reduce
km_tarjeta = reduce(lambda acum, c: acum + c["km"],
                    filter(lambda c: c["tarjeta"], carreras), 0)

print(f"Recaudación total: {total} €")
print(f"Carrera más larga: {mas_larga['pasajero']}")
print(f"Km pagados con tarjeta: {km_tarjeta}")
````

### Ejercicio 3.6 · Funciones que devuelven funciones
`crear_tarifa(precio_km)` devuelve una función que recibe km y devuelve el precio.

````python
def crear_tarifa(precio_km):
    def calcular(km):
        return round(km * precio_km, 2)
    return calcular

tarifa_diurna = crear_tarifa(1.10)
tarifa_nocturna = crear_tarifa(1.35)

print(tarifa_diurna(10))
print(tarifa_nocturna(10))
print([tarifa_nocturna(c["km"]) for c in carreras])
````

### Ejercicio 4.1 · Sumar los dígitos de un código

````python
def suma_digitos(n):
    if n < 10:                 # caso base: un solo dígito
        return n
    return n % 10 + suma_digitos(n // 10)

print(suma_digitos(4528))
print(suma_digitos(7))
````

### Ejercicio 4.2 · Total de una lista de gastos

````python
def total_gastos(gastos):
    if not gastos:             # caso base: lista vacía
        return 0
    return gastos[0] + total_gastos(gastos[1:])

print(total_gastos([12.5, 30, 7.5, 50]))
print(total_gastos([]))
````

### Ejercicio 4.3 · ¿Es un palíndromo?

````python
def es_palindromo(texto):
    if len(texto) <= 1:        # caso base: 0 o 1 caracteres siempre es palíndromo
        return True
    if texto[0] != texto[-1]:  # caso base: los extremos no coinciden
        return False
    return es_palindromo(texto[1:-1])

print(es_palindromo("reconocer"))
print(es_palindromo("radar"))
print(es_palindromo("python"))
````

### Ejercicio 4.4 · Cajas dentro de cajas

````python
envio = [2, [3, 1], [[4, 5], 2], [], 6]

def peso_total(caja):
    total = 0
    for elemento in caja:
        if isinstance(elemento, list):   # es una caja → llamada recursiva
            total += peso_total(elemento)
        else:                            # es un artículo → sumo su peso
            total += elemento
    return total

def contar_articulos(caja):
    total = 0
    for elemento in caja:
        if isinstance(elemento, list):
            total += contar_articulos(elemento)
        else:
            total += 1
    return total

print(f"Peso total: {peso_total(envio)} kg")
print(f"Número de artículos: {contar_articulos(envio)}")
````

> Aquí el caso base es implícito: una caja vacía `[]` no entra en el bucle y devuelve `0`.

### Ejercicio 4.5 · Iterativo frente a recursivo

````python
def ahorro_iterativo(capital, anios):
    for _ in range(anios):
        capital = capital * 1.02
    return capital

def ahorro_recursivo(capital, anios):
    if anios == 0:
        return capital
    return ahorro_recursivo(capital * 1.02, anios - 1)

print(round(ahorro_iterativo(1000, 5), 2))
print(round(ahorro_recursivo(1000, 5), 2))
````

**Pregunta:** con `anios = 5000` la versión recursiva necesita 5000 llamadas anidadas y Python tiene un límite por defecto de unas **1000** (`sys.getrecursionlimit()`), así que lanza **`RecursionError`**. La versión iterativa no tiene ese problema. Lo comprobamos:

````python
import sys
print("Límite de recursión:", sys.getrecursionlimit())
print("Iterativo 5000 años:", ahorro_iterativo(1000, 5000))
try:
    ahorro_recursivo(1000, 5000)
except RecursionError as e:
    print("Recursivo 5000 años → RecursionError:", e)
````

### Ejercicio 5.1 · Artículos de una tienda de deportes

````python
class Articulo:
    def __init__(self, nombre, precio, stock=0):
        self.nombre = nombre
        self.precio = precio
        self.stock = stock

    def valor_inventario(self):
        return self.precio * self.stock

    def vender(self, cantidad):
        if cantidad <= self.stock:
            self.stock -= cantidad
            return True
        return False

    def reponer(self, cantidad):
        self.stock += cantidad

    def __str__(self):
        return f"{self.nombre} | {self.precio} € | {self.stock} uds."


a1 = Articulo("Balón de fútbol", 25.0, 10)
a2 = Articulo("Cinta de pádel", 9.5)

print(a1)
print(a2)
print(a1.vender(3))
print(a2.vender(1))
a2.reponer(20)
print(a1)
print(a2)
print(a1.valor_inventario(), a2.valor_inventario())
````

### Ejercicio 5.2 · Cuentas de un banco

````python
class Cuenta:
    total_cuentas = 0          # atributo de clase
    COMISION_RETIRADA = 1.5    # atributo de clase (constante)

    def __init__(self, titular, saldo=0.0):
        self.titular = titular
        self._saldo = saldo
        Cuenta.total_cuentas += 1   # ¡con el nombre de la clase, no con self!

    def ingresar(self, importe):
        if importe > 0:
            self._saldo += importe

    def retirar(self, importe):
        total = importe + Cuenta.COMISION_RETIRADA
        if total <= self._saldo:
            self._saldo -= total
            return True
        return False

    def consultar_saldo(self):
        return self._saldo

    @classmethod
    def cuentas_creadas(cls):
        return cls.total_cuentas

    def __repr__(self):
        return f"Cuenta({self.titular!r}, {self._saldo})"


c1 = Cuenta("Laura", 100.0)
c2 = Cuenta("Diego")

c1.ingresar(50)
c1.ingresar(-20)
print(c1.retirar(48.5))
print(c2.retirar(10))
print(c1, c2)
print("Cuentas creadas:", Cuenta.cuentas_creadas())
````

> `{self.titular!r}` usa `repr()` del texto, que añade las comillas: `'Laura'`.
> `print()` usa `__str__`; como no lo hemos definido, Python usa `__repr__`.

### Ejercicio 5.3 · Carrito de la compra (composición)

````python
class Carrito:
    def __init__(self):
        self.lineas = []          # el carrito CONTIENE líneas (articulo, cantidad)

    def agregar(self, articulo, cantidad):
        self.lineas.append((articulo, cantidad))

    def total(self):
        return sum(articulo.precio * cantidad for articulo, cantidad in self.lineas)

    def resumen(self):
        for articulo, cantidad in self.lineas:
            print(f"{cantidad} x {articulo.nombre} = {articulo.precio * cantidad} €")
        print(f"TOTAL: {self.total()} €")


carrito = Carrito()
carrito.agregar(Articulo("Balón de fútbol", 25.0, 10), 3)
carrito.agregar(Articulo("Cinta de pádel", 9.5, 20), 2)
carrito.agregar(Articulo("Botella deportiva", 15.0, 5), 1)
carrito.resumen()
````

## Sección 6. Sobrecarga de operadores y conversión de tipos
### Ejercicio 6.1 · Sumar y comparar importes

````python
class Importe:
    def __init__(self, cantidad):
        self.cantidad = cantidad

    def __add__(self, otro):            # a + b
        return Importe(self.cantidad + otro.cantidad)

    def __eq__(self, otro):             # a == b
        # se redondea porque 19.99 + 5.01 en coma flotante puede no dar 25 exacto
        return round(self.cantidad, 2) == round(otro.cantidad, 2)

    def __lt__(self, otro):             # a < b
        return self.cantidad < otro.cantidad

    def __str__(self):                  # str(a) / print(a)
        return f"{self.cantidad:.2f} €"

    def __float__(self):                # float(a)
        return float(self.cantidad)


a = Importe(19.99)
b = Importe(5.01)
c = Importe(25)

total = a + b
print(total)
print(total == c)
print(a < b)
print(float(a) * 2)
````

### Ejercicio 6.2 · Lista de reproducción

````python
class ListaReproduccion:
    def __init__(self, nombre):
        self.nombre = nombre
        self._canciones = []

    def __len__(self):                  # len(lista)
        return len(self._canciones)

    def __contains__(self, titulo):     # "titulo" in lista
        return titulo in self._canciones

    def __iadd__(self, titulo):         # lista += "titulo"
        self._canciones.append(titulo)
        return self                     # ¡imprescindible! si no, viaje pasaría a valer None

    def __str__(self):
        return f"{self.nombre} ({len(self)} canciones)"


viaje = ListaReproduccion("Viaje")
viaje += "Canción A"
viaje += "Canción B"
viaje += "Canción C"

print(viaje)
print(len(viaje))
print("Canción B" in viaje)
print("Canción Z" in viaje)
````

### Ejercicio 6.3 · Precios escritos a mano

````python
class Precio:
    def __init__(self, euros):
        self.euros = float(euros)

    @staticmethod
    def desde_texto(texto):
        # "12,50 €" → "12,50" → "12.50" → 12.5
        limpio = texto.replace("€", "").strip().replace(",", ".")
        return Precio(float(limpio))

    def __float__(self):
        return self.euros

    def __bool__(self):
        return self.euros != 0

    def __str__(self):
        return f"{self.euros:.2f} €"


carta = ["12,50 €", "0,00 €", "7,25 €", "18,00 €"]
precios = [Precio.desde_texto(t) for t in carta]

print([float(p) for p in precios])
print("Total carta:", sum(float(p) for p in precios))
print("Platos gratuitos:", len([p for p in precios if not p]))
print(precios[2])
````

> **`@staticmethod` vs `@classmethod`:** el estático no recibe ni `self` ni `cls`. Aquí también valdría `@classmethod` con `return cls(...)`, que tiene la ventaja de funcionar bien con subclases.

## Sección 7. Herencia, polimorfismo y herencia múltiple
### Ejercicio 7.1 · Peajes de una autopista

````python
class Vehiculo:
    def __init__(self, matricula, kilometros):
        self.matricula = matricula
        self.kilometros = kilometros

    def peaje(self):
        return self.kilometros * 0.10


class Moto(Vehiculo):
    def peaje(self):
        return super().peaje() / 2


class Camion(Vehiculo):
    def __init__(self, matricula, kilometros, ejes):
        super().__init__(matricula, kilometros)
        self.ejes = ejes

    def peaje(self):
        return super().peaje() + 2 * self.ejes


vehiculos = [
    Vehiculo("1234-ABC", 100),
    Moto("5678-DEF", 100),
    Camion("9012-GHI", 100, 4),
]

total = 0
for v in vehiculos:
    importe = v.peaje()           # ← polimorfismo
    print(f"{v.matricula} → {round(importe, 2)} €")
    total += importe

print(f"Total recaudado: {round(total, 2)} €")
````

**Pregunta:** el polimorfismo está en la línea `v.peaje()` dentro del bucle: se escribe **una sola llamada** y cada objeto ejecuta **su propia versión** del método según su clase, sin ningún `if` que pregunte por el tipo.

### Ejercicio 7.2 · Suscripciones de una plataforma de series

````python
class Suscripcion:
    def __init__(self, usuario, cuota_base):
        self.usuario = usuario
        self.cuota_base = cuota_base

    def precio_mensual(self):
        return self.cuota_base

    def ficha(self):
        return f"{self.usuario} ({type(self).__name__}): {self.precio_mensual():.2f} €"


class Familiar(Suscripcion):
    def __init__(self, usuario, cuota_base, perfiles_extra):
        super().__init__(usuario, cuota_base)
        self.perfiles_extra = perfiles_extra

    def precio_mensual(self):
        return super().precio_mensual() + 3 * self.perfiles_extra


class Anual(Suscripcion):
    def precio_mensual(self):
        return super().precio_mensual() * 0.85


class ConAnuncios(Suscripcion):
    def precio_mensual(self):
        return super().precio_mensual() * 0.60


suscripciones = [
    Suscripcion("Carlos", 12),
    Familiar("Lucía", 12, 2),
    Anual("Sergio", 12),
    ConAnuncios("Nerea", 12),
]

for s in suscripciones:
    print(s.ficha())

ingresos = sum(s.precio_mensual() for s in suscripciones)
print(f"Ingresos mensuales: {ingresos:.2f} €")
````

> `ficha()` está escrita **solo en la superclase**, pero llama a `self.precio_mensual()`, que es el de cada subclase. Y `type(self).__name__` da el nombre de la clase real del objeto.

### Ejercicio 7.3 · Impresora multifunción (herencia múltiple)

````python
class EquipoOficina:
    def __init__(self, modelo):
        self.modelo = modelo

    def encender(self):
        print(f"{self.modelo}: encendido")

    def descripcion(self):
        return "Equipo de oficina"


class Impresora(EquipoOficina):
    def imprimir(self, documento):
        print(f"Imprimiendo {documento}")

    def descripcion(self):
        return super().descripcion() + " + Impresora"


class Escaner(EquipoOficina):
    def escanear(self, documento):
        print(f"Escaneando {documento}")

    def descripcion(self):
        return super().descripcion() + " + Escáner"


class Multifuncion(Impresora, Escaner):
    def fotocopiar(self, documento):
        self.escanear(documento)
        self.imprimir(documento)

    def descripcion(self):
        return super().descripcion() + " + Multifunción"


equipo = Multifuncion("OfiPro 300")
equipo.encender()
equipo.fotocopiar("contrato.pdf")
print(equipo.descripcion())
print([clase.__name__ for clase in Multifuncion.__mro__])
````

**Pregunta:** el MRO es `Multifuncion → Impresora → Escaner → EquipoOficina → object`. Cada `descripcion()` llama **primero** a `super()` y **después** añade su texto. Las llamadas bajan por el MRO hasta `EquipoOficina` (que no llama a nadie y devuelve `"Equipo de oficina"`) y los textos se van añadiendo **a la vuelta**, en orden inverso al MRO: primero `Escaner`, luego `Impresora` y al final `Multifunción`. Por eso sale `Equipo de oficina + Escáner + Impresora + Multifunción`.

### Ejercicio 7.4 · Permisos de usuarios en una aplicación
**Parte A y B**

````python
class Usuario:
    def __init__(self, nombre):
        self.nombre = nombre
        print("Configurando Usuario")

    def permisos(self):
        return ["leer"]

    def puede(self, accion):
        return accion in self.permisos()


class Editor(Usuario):
    def __init__(self, nombre):
        super().__init__(nombre)
        print("Configurando Editor")

    def permisos(self):
        return super().permisos() + ["editar"]


class Moderador(Usuario):
    def __init__(self, nombre):
        super().__init__(nombre)
        print("Configurando Moderador")

    def permisos(self):
        return super().permisos() + ["borrar comentarios"]


class Administrador(Editor, Moderador):
    def __init__(self, nombre):
        super().__init__(nombre)
        print("Configurando Administrador")

    def permisos(self):
        return super().permisos() + ["gestionar usuarios"]


admin = Administrador("Marina")
print(admin.permisos())
print(admin.puede("borrar comentarios"), admin.puede("borrar la base de datos"))
print([clase.__name__ for clase in Administrador.__mro__])
````

**Parte C. Preguntas**

**1.** Porque `super()` sigue el MRO del objeto (`Administrador → Editor → Moderador → Usuario`), no "el padre de cada clase". El `super()` de `Editor` llama a `Moderador`, y solo el de `Moderador` llega a `Usuario`. Cada clase aparece una vez en el MRO, así que cada constructor se ejecuta una sola vez. Los mensajes salen "al revés" porque cada constructor imprime **después** de llamar a `super()`.

**2.** El administrador pierde **`"borrar comentarios"`**. Al no usar `super()`, `Editor` corta la cadena: nunca se llega a `Moderador.permisos()`, que es la siguiente en el MRO. Lo comprobamos:

````python
class EditorSinSuper(Usuario):
    def permisos(self):
        return ["leer", "editar"]          # sin super(): corta la cadena

class AdminPrueba(EditorSinSuper, Moderador):
    def permisos(self):
        return super().permisos() + ["gestionar usuarios"]

print(AdminPrueba("Prueba").permisos())
````

**3.** Con `class Administrador(Moderador, Editor)` el MRO pasa a ser `Administrador → Moderador → Editor → Usuario → object`. Cambian:
- el orden de los mensajes: `Usuario`, **`Editor`**, **`Moderador`**, `Administrador`;
- el orden de los permisos: `['leer', 'editar', 'borrar comentarios', 'gestionar usuarios']`;
- el MRO mostrado.

`puede()` sigue dando `True False`. Lo comprobamos:

````python
class Administrador2(Moderador, Editor):
    def __init__(self, nombre):
        super().__init__(nombre)
        print("Configurando Administrador")

    def permisos(self):
        return super().permisos() + ["gestionar usuarios"]

admin2 = Administrador2("Marina")
print(admin2.permisos())
print(admin2.puede("borrar comentarios"), admin2.puede("borrar la base de datos"))
print([clase.__name__ for clase in Administrador2.__mro__])
````

### Partes B, C y D. Superclase, subclases y clase `Agencia`

````python
class Reserva:
    def __init__(self, cliente, destino, cantidad, precio_unitario):
        self.cliente = cliente
        self.destino = destino
        self.cantidad = cantidad
        self.precio_unitario = precio_unitario

    def precio_base(self):
        return self.cantidad * self.precio_unitario

    def precio_total(self):
        return self.precio_base()

    def unidad(self):
        return "unidades"

    def tipo(self):
        return type(self).__name__.replace("Reserva", "").upper()

    def __str__(self):
        return (f"{self.tipo()} | {self.cliente} | {self.destino} | "
                f"{self.cantidad} {self.unidad()} | {round(self.precio_total(), 2)} €")


class ReservaVuelo(Reserva):
    TASAS = 25
    SUPLEMENTO_LARGO = 30          # por pasajero si el billete supera 200 €

    def unidad(self):
        return "pasajeros"

    def precio_total(self):
        precio = self.precio_base() + self.TASAS
        if self.precio_unitario > 200:
            precio += self.cantidad * self.SUPLEMENTO_LARGO
        return precio


class ReservaHotel(Reserva):
    TASA_TURISTICA = 1.5           # por noche

    def unidad(self):
        return "noches"

    def precio_total(self):
        precio = self.precio_base()
        if self.cantidad >= 3:
            precio *= 0.9
        return precio + self.cantidad * self.TASA_TURISTICA


class ReservaTren(Reserva):
    # No sobrescribe precio_total(): le vale el de la superclase
    def unidad(self):
        return "billetes"


class Agencia:
    def __init__(self):
        self.reservas = []

    def agregar(self, reserva):
        self.reservas.append(reserva)

    def mostrar(self):
        for reserva in self.reservas:
            print(reserva)

    def total(self):
        return sum(r.precio_total() for r in self.reservas)

    def contar_superiores_a(self, importe):
        return len([r for r in self.reservas if r.precio_total() > importe])
````

> Los `if` que quedan dentro de `ReservaVuelo` y `ReservaHotel` **no preguntan por el tipo** de reserva: son reglas de negocio de ese tipo concreto, así que cumplen las restricciones.

### Parte E. Programa principal

````python
agencia = Agencia()
agencia.agregar(ReservaVuelo("Ana López", "Madrid-Roma", 2, 120.0))
agencia.agregar(ReservaHotel("Luis Pérez", "Hotel Mar Azul", 3, 85.0))
agencia.agregar(ReservaTren("Marta Gil", "Madrid-Sevilla", 1, 45.0))
agencia.agregar(ReservaVuelo("Pablo Ruiz", "Madrid-Nueva York", 1, 300.0))
agencia.agregar(ReservaHotel("Eva Sanz", "Hotel Centro", 2, 140.0))

agencia.mostrar()
print(f"TOTAL AGENCIA: {round(agencia.total(), 2)} €")
print(f"Reservas de más de 200 €: {agencia.contar_superiores_a(200)}")
````

> En el enunciado los `f-string` aparecen sin espacio tras los dos puntos (`TOTAL AGENCIA:{...}`); lo he puesto con espacio para que la salida sea **idéntica** a la del programa original.

## Sección 9. Ejercicio integrador: restaurante de comida a domicilio
### Parte A. Jerarquía de clases

````python
from functools import reduce

class Plato:
    def __init__(self, nombre, precio):
        self.nombre = nombre
        self.precio = precio

    def precio_final(self):
        return self.precio

    def __str__(self):
        return f"{self.nombre} - {self.precio_final():.2f} €"


class PlatoVegano(Plato):
    def precio_final(self):
        return super().precio_final() * 0.90


class PlatoGourmet(Plato):
    RECARGO = 3

    def precio_final(self):
        return super().precio_final() + self.RECARGO
````

### Parte B. Datos del día

````python
pedidos = [
    {"cliente": "Ana",   "plato": PlatoVegano("Ensalada quinoa", 11.0),  "unidades": 2},
    {"cliente": "Luis",  "plato": Plato("Hamburguesa", 9.5),             "unidades": 1},
    {"cliente": "Marta", "plato": PlatoGourmet("Risotto trufa", 18.0),   "unidades": 1},
    {"cliente": "Ana",   "plato": Plato("Pizza margarita", 10.0),        "unidades": 3},
    {"cliente": "Pablo", "plato": PlatoVegano("Curry garbanzos", 12.0),  "unidades": 1},
]

def importe(pedido):
    """Importe de un pedido: precio final del plato × unidades."""
    return pedido["plato"].precio_final() * pedido["unidades"]
````

### Parte C. Comprehensions

````python
importes = [round(importe(p), 2) for p in pedidos]
clientes = {p["cliente"] for p in pedidos}
precios_finales = {p["plato"].nombre: round(p["plato"].precio_final(), 2) for p in pedidos}

print(importes)
print(sorted(clientes))
print(precios_finales)
````

### Parte D. Programación funcional

````python
# 1. filter + isinstance
veganos = list(filter(lambda p: isinstance(p["plato"], PlatoVegano), pedidos))

# 2. map + reduce
facturacion = reduce(lambda a, b: a + b, map(importe, pedidos), 0)

# 3. Función que devuelve una función
def crear_filtro_minimo(minimo):
    def filtro(pedido):
        return importe(pedido) >= minimo
    return filtro

minimo_20 = crear_filtro_minimo(20)
pedidos_20 = list(filter(minimo_20, pedidos))

print([p["plato"].nombre for p in veganos])
print(round(facturacion, 2))
print([p["cliente"] for p in pedidos_20])
````

### Parte E. Recursión

````python
def contar_unidades(pedidos):
    if not pedidos:                     # caso base: no quedan pedidos
        return 0
    return pedidos[0]["unidades"] + contar_unidades(pedidos[1:])

print(contar_unidades(pedidos))
````

### Parte F. Informe final

````python
print("=== CARTA DEL DÍA ===")
for p in pedidos:
    print(p["plato"])
print()
print("Importes por pedido:", importes)
print("Clientes distintos:", sorted(clientes))
print("Precios finales:", precios_finales)
print("Platos veganos pedidos:", [p["plato"].nombre for p in veganos])
print(f"Facturación total: {round(facturacion, 2)} €")
print("Pedidos de 20 € o más:", [p["cliente"] for p in pedidos_20])
print("Unidades vendidas:", contar_unidades(pedidos))
````

**Pregunta final:** solo habría que **añadir** la clase `PlatoInfantil` (sobrescribiendo `precio_final()` para que devuelva 6) y usarla en los datos. Todo lo demás funciona sin tocarlo porque trabaja con el método `precio_final()` y no pregunta por el tipo: `__str__`, la carta, los importes, el diccionario de precios, la facturación, el filtro de mínimo y `contar_unidades`.

La única parte que depende de un tipo concreto es el filtro de **veganos** (`isinstance(..., PlatoVegano)`): sigue funcionando, pero si se quisiera un informe de "platos infantiles" habría que escribir otro filtro.

### Ejercicio 10.1 · Crear e importar un módulo
Crea un módulo `utilidades_tienda.py` con una constante `IVA = 0.21` y las funciones `precio_con_iva(precio)` y `formatear_euros(cantidad)`. Impórtalo de tres formas distintas.

````python
%%writefile utilidades_tienda.py
"""Funciones de ayuda para la tienda."""

IVA = 0.21

def precio_con_iva(precio):
    return round(precio * (1 + IVA), 2)

def formatear_euros(cantidad):
    return f"{cantidad:.2f} €"

if __name__ == "__main__":
    # Solo se ejecuta si lanzamos el archivo directamente: python utilidades_tienda.py
    print("Probando el módulo:", formatear_euros(precio_con_iva(10)))
````

- **¿Qué inconveniente tiene `from utilidades_tienda import *`?** Trae todos los nombres sin prefijo y puede **pisar** variables o funciones nuestras con el mismo nombre sin que nos demos cuenta; además no se sabe de dónde viene cada nombre.
- **Si importo el módulo dos veces, ¿se ejecuta dos veces?** No: Python lo guarda en `sys.modules` y la segunda vez reutiliza el ya cargado. Para recargarlo tras modificarlo, `importlib.reload(modulo)`.

````python
%%writefile agencia/reservas.py
class Reserva:
    def __init__(self, cliente, cantidad, precio_unitario):
        self.cliente = cliente
        self.cantidad = cantidad
        self.precio_unitario = precio_unitario

    def precio_total(self):
        return self.cantidad * self.precio_unitario


class ReservaTren(Reserva):
    pass


class ReservaHotel(Reserva):
    def precio_total(self):
        return super().precio_total() + self.cantidad * 1.5
````

````python
%%writefile agencia/informes.py
from .reservas import Reserva      # import relativo dentro del paquete

def total(reservas):
    return sum(r.precio_total() for r in reservas)

def mas_cara(reservas):
    return max(reservas, key=lambda r: r.precio_total())
````

````python
from agencia import ReservaTren, ReservaHotel   # gracias a __init__.py
from agencia import informes                    # submódulo
from agencia.informes import mas_cara           # función concreta de un submódulo

lista = [ReservaTren("Marta", 1, 45.0), ReservaHotel("Eva", 2, 140.0)]
print("Total:", informes.total(lista))
print("Más cara:", mas_cara(lista).cliente)
````

**Preguntas:**
- **¿Para qué sirve `__init__.py`?** Marca la carpeta como paquete y se ejecuta al importarlo. Sirve para inicializarlo y para "subir" nombres (por eso podemos escribir `from agencia import ReservaTren` en vez de `from agencia.reservas import ReservaTren`).
- **¿Qué significa el punto en `from .reservas import Reserva`?** Es un **import relativo**: "el módulo `reservas` del mismo paquete".
- **¿Dónde busca Python los módulos?** En la lista `sys.path`: primero la carpeta del script/cuaderno, luego las rutas del sistema y los paquetes instalados.

### Ejercicio 10.3 · Módulos de la biblioteca estándar
Usa `math`, `random` y `datetime` para:
1. Calcular la raíz cuadrada de 144 y redondear hacia arriba 7.2.
2. Elegir un ganador al azar de una lista (con semilla para que el resultado sea reproducible).
3. Mostrar la fecha de hoy con el formato `dd/mm/aaaa` y cuántos días faltan para el 31 de diciembre.

````python
import math
import random
from datetime import date

print(math.sqrt(144), math.ceil(7.2))

random.seed(42)
print("Ganador:", random.choice(["Ana", "Luis", "Marta", "Pablo"]))

hoy = date.today()
fin_de_anio = date(hoy.year, 12, 31)
print("Hoy:", hoy.strftime("%d/%m/%Y"))
print("Días hasta fin de año:", (fin_de_anio - hoy).days)
````

### Ejercicio 10.4 · Lectura de código: ¿qué se muestra?

````python
%%writefile contador.py
print("Cargando contador...")
visitas = 0

def visitar():
    global visitas
    visitas += 1
    return visitas
````

````python
import sys
sys.modules.pop("contador", None)   # (solo para que el ejercicio se pueda repetir en Jupyter)

import contador
import contador
from contador import visitar

visitar()
visitar()
print(contador.visitas)
````

**Salida:**
```
Cargando contador...
2
```
- `"Cargando contador..."` sale **una sola vez**: el código del módulo se ejecuta solo en el primer `import`; los siguientes reutilizan el módulo ya cargado.
- `from contador import visitar` no vuelve a cargar nada; `visitar` es la misma función del módulo y modifica `contador.visitas`, que acaba valiendo `2`.
