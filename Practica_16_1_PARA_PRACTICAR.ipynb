````python
# ⚙️ Celda de configuración: ejecútala siempre la primera
import io, contextlib, textwrap

def _normalizar(texto):
    return "\n".join(l.strip() for l in textwrap.dedent(texto).strip().splitlines() if l.strip())

def comprobar(prueba, esperado):
    buffer = io.StringIO()
    try:
        with contextlib.redirect_stdout(buffer):
            prueba()
    except Exception as e:
        print(buffer.getvalue())
        print(f"❌ Tu código lanza un error: {type(e).__name__}: {e}")
        return
    obtenido = _normalizar(buffer.getvalue())
    if obtenido == _normalizar(esperado):
        print("✅ ¡Correcto!\n")
        print(obtenido)
    else:
        print("❌ Todavía no. Compara línea a línea:\n")
        print("--- Lo que sale ---")
        print(obtenido or "(nada)")
        print("\n--- Lo esperado ---")
        print(_normalizar(esperado) or "(no has escrito tu predicción)")

print("Listo para practicar 💪")
````

2. `super()` = "la **siguiente** clase del MRO", no "mi padre".
3. Si el método hace `super()...` **y después** añade algo, los añadidos salen en **orden inverso** al MRO (se añaden "a la vuelta").

| Operación | Método |
|---|---|
| `a + b` | `__add__(self, otro)` |
